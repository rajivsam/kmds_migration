# Clean Olist Dataset

This notebook loads the cleaned order-level dataset produced by the KMDS dd-parser-cleaner workflow and displays the associated metadata for the current featurization input.

A data quality issue surfaced in this dataset: some orders are missing `order_item_id` or `price`. Those rows are dropped before featurization.

The dataset queried here is the cleaned dataset used by the featurization pipeline.


```python
import logging
import pandas as pd
from pathlib import Path

logging.basicConfig(level=logging.INFO, format='%(asctime)s %(levelname)s %(message)s')
clean_path = Path("../data/dd_cleaner/olist_daily_orders_prepared_clean.csv")
metadata_path = Path("../data_dictionary/olist_example_dd.csv")

if not clean_path.exists():
    raise FileNotFoundError(f"Cleaned dataset not found: {clean_path}")
if not metadata_path.exists():
    raise FileNotFoundError(f"Metadata dictionary not found: {metadata_path}")

s_raw = pd.read_csv(clean_path, parse_dates=["order_purchase_timestamp"], low_memory=False)
df_metadata = pd.read_csv(metadata_path, low_memory=False)

mask = s_raw["order_item_id"].notna() & s_raw["price"].notna()
dropped = len(s_raw) - int(mask.sum())
logging.info(f"Dropped {dropped} rows missing order_item_id or price from the cleaned dataset.")
df_raw = s_raw[mask].copy()

print("Cleaned dataset path:", clean_path)
print("Raw dataset shape after dropping incomplete rows:", df_raw.shape)
print("Metadata dictionary path:", metadata_path)
print("Metadata rows:", len(df_metadata))
print("Metadata columns:", df_metadata.columns.tolist())

print("Sample metadata records:")
print(df_metadata.head(10).to_string(index=False))
print("Sample cleaned dataset rows:")
print(df_raw.head(5).to_string(index=False))
```
