# KMDS Clustering Advisor for Olist SP 2017
This notebook runs the `kmds_modeling` advisor for the clustering task using the featurization output and the local rationale for spectral clustering.

```python
import sys
from pathlib import Path
import importlib.util

import kmds_modeling

workspace_root = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
workspace_root = workspace_root.resolve()

config_path = workspace_root / 'modeling_config.yaml'
if not config_path.exists():
    raise FileNotFoundError(f'Model config not found: {config_path}')

advisor_module_path = workspace_root / 'models' / 'clustering_advisor.py'
if not advisor_module_path.exists():
    raise FileNotFoundError(f'Advisor module not found: {advisor_module_path}')

spec = importlib.util.spec_from_file_location('clustering_advisor', advisor_module_path)
if spec is None or spec.loader is None:
    raise ImportError(f'Unable to load module spec from {advisor_module_path}')

clustering_advisor = importlib.util.module_from_spec(spec)
spec.loader.exec_module(clustering_advisor)
run_clustering_advisor = clustering_advisor.run_clustering_advisor

print('kmds_modeling version:', kmds_modeling.get_package_info().get('version'))

advisor_output = run_clustering_advisor(str(config_path))
print('Advisor recommendation path:', advisor_output['output_path'])
print('Profile:')
print(advisor_output['profile'])
print('\nRecommendation:')
print(advisor_output['recommendation'])
```

## Advisor Rationale
The advisor confirms that this use case is a clustering task. Based on the product-week matrix structure and the attached domain rationale, spectral clustering is preferred because it respects the bipartite week-product topology and the joint dual-manifold structure.
Experts on the data science team believe that spectral co-clustering better captures seasonal shopping patterns and product affinities for SP 2017 than flat k-means on the raw high-dimensional product matrix.
