# Clean California Housing Dataset

This notebook mirrors the KMDS cleaning workflow used in the SBA migration example. It loads the project workspace, prints the available artifacts, and surfaces the raw dataset, tagged entities, and metadata table for review and cleaning.

Run this notebook from either the project root or the `notebooks/` directory. The code resolves the workspace automatically.


```python
from pathlib import Path

from dd_cleaner.notebook_utils import (
    get_cleaned_data,
    get_metadata_table,
    get_raw_data,
    get_tagged_entities,
    init_notebook_session,
)

NOTEBOOK_DIR = Path.cwd().resolve()
WORKING_DIR = NOTEBOOK_DIR if (NOTEBOOK_DIR / "config.yaml").exists() else NOTEBOOK_DIR.parent
WORKING_DIR = WORKING_DIR.resolve()

coord, artifacts_df = init_notebook_session(str(WORKING_DIR))
artifacts_df
```

```python
print(f"Workspace root: {WORKING_DIR}")
print(f"Raw dataset path: {coord.raw_dataset_path}")
print(f"Tagged dictionary path: {coord.data_dictionary_csv_path}")
print(f"Metadata table path: {coord.metadata_table_path}")
```

```python
try:
    df_raw = get_raw_data(coord)
    print(f"Rows: {len(df_raw)} | Columns: {len(df_raw.columns)}")
    df_raw.head()
except FileNotFoundError as exc:
    print(exc)
```

```python
try:
    df_tags = get_tagged_entities(coord)
    print(f"Tagged attributes: {len(df_tags)}")
    df_tags.head()
except FileNotFoundError as exc:
    print(exc)
```

```python
try:
    df_cleaned = get_cleaned_data(coord)
    print(f"Cleaned rows: {len(df_cleaned)} | Cleaned columns: {len(df_cleaned.columns)}")
    df_cleaned.head()
except FileNotFoundError as exc:
    print(exc)
```

```python
try:
    df_metadata = get_metadata_table(coord)
    print(f"Metadata rows: {len(df_metadata)}")
    df_metadata.head(20)
except FileNotFoundError as exc:
    print(exc)
```

```python
# Optional quick profile for the metadata and tag dimensions
for label, frame in {
    "raw": df_raw if "df_raw" in globals() else None,
    "tags": df_tags if "df_tags" in globals() else None,
    "metadata": df_metadata if "df_metadata" in globals() else None,
}.items():
    if frame is not None:
        print(f"\n--- {label} ---")
        print(frame.dtypes)
        print(frame.isna().sum().sort_values(ascending=False).head(10).to_string())
```
