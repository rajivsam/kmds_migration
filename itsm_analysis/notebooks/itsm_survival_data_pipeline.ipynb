# Verify ITSM Survival Featurization Pipeline
This notebook cell runs the new survival featurization pipeline and checks the generated outputs.

```python
import os
import pandas as pd
from pathlib import Path
from featurization.core.sequential_pipeline_runner import PipelineRunner

workspace_root = Path(os.getcwd()).parent
print(f'Workspace root: {workspace_root}')
runner = PipelineRunner(str(workspace_root))
df = runner.accumulate_stages()
print('Pipeline output shape:', df.shape)
display(df.head())
resolver = runner.context['resolver']
print('Featurized dataset path:', resolver.featurized_dataset_path)
print('Model-ready dataset path:', resolver.model_ready_dataset_path)

output_dir = Path(workspace_root) / "data" / resolver.config.get("featurization_output_dir", "featurization")
km_path = output_dir / resolver.config.get("itsm_KM_data_file", "itsm_KM_data.csv")
pr_path = output_dir / resolver.config.get("itsm_PR_data_file", "itsm_PR_data.csv")

print('\nLoading resulting split datasets:')
print('KM dataset path:', km_path)
print('PR dataset path:', pr_path)

km_df = pd.read_csv(km_path)
pr_df = pd.read_csv(pr_path)

print('\nKaplan-Meier dataset sample:')
display(km_df.head())
print('\nParametric regression dataset sample:')
display(pr_df.head())
```
